# Método da Bissecção

Implementação do Método da Bissecção, contextualizado em exemplos.

## Exemplo 1: Resolver $\cos(x) = x$.

```python
import numpy as np

def f(x):
    y = np.cos(x) - x
    return y
```

```python
a = 0
b = np.pi/2

N = 10

k = 0
while k <= 20:
    m = (a+b)/2

    y = f(m)

    if y == 0:
        print(f"Solução exata: {m}")
    else:
        if y*f(a)<0:

            b = m
            k = k+1
        else:
            a = m
            k = k + 1

print(f"Solução aproximada: {m}")
```
