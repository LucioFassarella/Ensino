# **Limites**: Determinação da precisão $\delta > 0 $ de variável em termos da tolerância $\varepsilon > 0$ para dado limite $L = \lim_{x \rightarrow x_0}f(x)$.

## Exemplo 1: $$f(x) = \frac{e^{2x} - 1}{x},\ x_0 = 0,\ L = 2;\ \varepsilon = 0.1$$

```python
import numpy as np

def f(x):
    y = (np.exp(2*x) - 1)/x
    return y
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate x values in the interval [-2, 2]
x_values = np.linspace(-0.04, 0.04, 400)

# Calculate y values using the function f(x)
y_values = f(x_values)

# Create the plot
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label='f(x)', color='blue')

# Add the horizontal line
plt.axhline(y= 2 + 0.1, color='red', linestyle='--', label='y = 1')
# Add the horizontal line
plt.axhline(y= 2 - 0.1, color='red', linestyle='--', label='y = 1')
# Add the horizontal line y = 1.5
plt.axhline(y=2, color='gray', linestyle='--', label='y = 1')

# Add the vertical line x = 1
plt.axvline(x=-0.03, color='green', linestyle=':', label='x = 1')
# Add the vertical line x = 1
plt.axvline(x=+0.03, color='green', linestyle=':', label='x = 1')
# Add the vertical line x = 0
plt.axvline(x=0, color='black', linestyle=':', label='x = 1')

# Add labels and title
plt.title('Gráfico de f(x) com retas y=1 e x=1')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```
